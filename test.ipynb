```python
import xarray as xr
ds = xr.open_dataset(path+'/fields.zarr')
```

```python
print(ds)
```

```python
out = path + '_test_only/fields.zarr'
print(out)
ds.to_zarr(out)
```

```python
forcings = xr.open_dataset(path+'/forcing.zarr')
```

```python
forcings = forcings.sel(time=forcings.time.dt.year > 2015)
```

```python
out = path + '_test_only/forcing.zarr'
print(out)
forcings.to_zarr(out, mode='w')
```

```python
print(forcings.data_vars)
```

```python
def clean_attrs(ds):
    ds = ds.copy()
    ds.attrs = {str(k): str(v) for k, v in ds.attrs.items()}
    for var in ds.variables:
        ds[var].attrs = {str(k): str(v) for k, v in ds[var].attrs.items()}
    return ds

forcings = clean_attrs(forcings)
forcings.to_zarr(out, mode="w")
```
