```python
import numpy as np

data = np.load("../shap_results.npz")
shap_values = data["shap_values"]
background_np = data["background"]
sample_np = data["samples"]

print(shap_values.shape)
```

```python
# shap_values has shape [batch, total_features]
shap_vals = shap_values[0]   # take first explained sample

# split init vs forcing
n_init = 2 * 8192 * 83
init_vals = shap_vals[:n_init].reshape(2, 8192, 83)
forcing_vals = shap_vals[n_init:].reshape(1, 8192, 15)
```

```python
init_importance = init_vals.mean(axis=(1,2))        # shape [2]
forcing_importance = forcing_vals.mean(axis=(1,2)) # shape [1]

all_importance = np.concatenate([init_importance, forcing_importance])
```

```python
init_importance = init_vals.mean(axis=(1,2))        # shape [2]
forcing_importance = forcing_vals.mean(axis=(1,2)) # shape [1]

all_importance = np.concatenate([init_importance, forcing_importance])
```

```python
import matplotlib.pyplot as plt

labels = ["init_state_t-1", "init_state_t", "forcing_0"]
plt.bar(labels, all_importance)
plt.ylabel("Mean |SHAP value|")
plt.title("Variable importance")
plt.show()
```
